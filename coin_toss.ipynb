** Coin Toss Experiment ** 

```python
from qiskit import QuantumCircuit

qcoin = QuantumCircuit(1)   
qcoin.h(0)  # Apply Hadamard gate to create superposition

qcoin.measure_all()  # Measure the qubit

qcoin.draw('mpl')  # Draw the circuit using matplotlib
```

```python
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram

sampler = StatevectorSampler()
nflip = 1000  # Number of coin tosses

pubs = ([qcoin])
job = sampler.run(pubs, shots=nflip)
results = job.result()
data_pub = results[0].data

counts = data_pub.meas.get_counts()
plot_histogram(counts)  # Plot the histogram of results

```

```python
from qiskit.primitives import StatevectorEstimator
from qiskit.quantum_info import Pauli, SparsePauliOp

qcoin = QuantumCircuit(1)
qcoin.h(0)  # Apply Hadamard gate to create superposition

estimator = StatevectorEstimator()
observables = [[Pauli('X')]]
pubs = (qcoin,observables)
job = estimator.run([pubs])

result = job.result()[0]

print(result.data.evs)
```

```python
qcoin_0 = QuantumCircuit(1)
qcoin_0.h(0)  # Apply Hadamard gate to create superposition
qcoin_0.h(0)  # Apply Hadamard gate again to return to |0>

qcoin_0.measure_all()  # Measure the qubit
qcoin_0.draw('mpl')  # Draw the circuit using matplotlib
```
